```python
import sys
import os
sys.path.insert(0, os.path.abspath("../src"))
print("Path configured.")
```

```python
from retrieval import get_or_build_vector_store
from rag_chain import build_rag_chain, ask_from_image, print_answer
from image_captioner import load_captioner, caption_image

print("Loading vector store...")
vs = get_or_build_vector_store()

print("Building RAG chain...")
chain, retriever = build_rag_chain(vs)

print("Loading captioner (once)...")
processor, model = load_captioner()

print("All loaded. Ready.")
```

```python
# Test captions before running the full pipeline
# so you can see what BLIP-2 actually "sees"
test_images = [
    "../data/test_images/robot.jpg",
    "../data/test_images/hand.jpg",
    "../data/test_images/mask.jpg",
    "../data/test_images/intel.jpg",
]

```

```python
print("=== Caption Quality Check ===\n")
for img_path in test_images:
    caption = caption_image(img_path, processor, model)
    print(f"Image  : {img_path.split('/')[-1]}")
    print(f"Caption: {caption}")
    print()
```

```python
import time

print("=== Full Pipeline Test ===\n")
for img_path in test_images:
    print(f"\nImage: {img_path.split('/')[-1]}")
    print("-" * 40)
    
    result = ask_from_image(img_path, chain, retriever, processor, model)
    
    print(f"Caption   : {result['caption']}")
    print(f"Top source: {result['sources'][0]['source'] if result['sources'] else 'None'}")
    print(f"Answer    : {result['answer'][:300]}...")
    
    time.sleep(3)
```
